**Nasıl kullanılır:**

**Analiz özeti:** AI ölçümü, 3 kör uzmanın ortalamasıyla karşılaştırılır. Birincil karar Bland-Altman %95 uyum sınırları (LoA) ile önceden belirlenen kabul sınırına göre verilir. İki seviye vardır:
- **nokta:** LoA kabul sınırının içinde
- **sıkı:** LoA'nın %95 güven aralığı da sınırın içinde

````python
# ---------------------------------------------------------------------------
# ÇALIŞMA KONFİGÜRASYONU
# Kabul sınırları literatürden türetildi (gerekçeler README.md'de):
#   Mesane: ±55 mL  -> manuel 2D US okuyucular arası uyum 3.2 ± 54.4 mL
#                      (Revolutionizing Bladder Health, 2024, PMC11353831);
#                      FDA K230497 Bladder AI sonucu −42/+46 mL bu sınırın içinde
#   Karaciğer: ±%15 -> FDA'nın kabul ettiği organ uzunluğu emsalleri
#                      (Philips K260667: böbrek ±%14.3, dalak ±%15.9);
#                      karaciğer okuyucular arası ICC 0.71–0.94 (Childs 2014)
# ---------------------------------------------------------------------------
CONFIG = {
    "bladder": {
        "title": "AI Mesane Hacmi",
        "unit": "mL",
        "readers": ["reader1_ml", "reader2_ml", "reader3_ml"],
        "ai": "ai_ml",
        "primary": {"name": "Hacim farkı (mL)", "scale": "abs", "limit": 55.0},
        "secondary": {"name": "Yüzde fark, referans ≥100 mL (%)", "scale": "pct",
                      "limit": None, "min_ref": 100.0},
        "cutoff": {"value": 200.0, "label": "PVR > 200 mL (artmış rezidü)"},
        "power": {"bias": 2.0, "sd": 22.0, "delta": 55.0},
    },
    "liver": {
        "title": "AI Karaciğer Kraniokaudal Uzunluğu (MCL)",
        "unit": "cm",
        "readers": ["reader1_cm", "reader2_cm", "reader3_cm"],
        "ai": "ai_cm",
        "primary": {"name": "Yüzde fark (%)", "scale": "pct", "limit": 15.0},
        "secondary": {"name": "Mutlak fark (cm)", "scale": "abs", "limit": None,
                      "min_ref": None},
        "cutoff": {"value": 16.0, "label": "Hepatomegali > 16 cm"},
        "power": {"bias": -1.5, "sd": 5.0, "delta": 15.0},
    },
}

SUBGROUP_COLS = ["site", "device_type", "sex", "bmi_category", "age_group", "image_quality"]

COL = {"ink": "#14213D", "accent": "#C2621B", "blue": "#2B6CB0", "grey": "#8A94A6",
       "band": "#E9EDF3", "pass": "#2F855A", "fail": "#C53030"}
````

## 2. İstatistik fonksiyonları

| Fonksiyon | Ne yapar |
|---|---|
| `differences` | Mutlak fark (AI − Ref) veya yüzde fark ((AI − Ref) / ortalama × 100) |
| `bland_altman` | Bias, SD, %95 LoA ve bunların güven aralıkları (Bland & Altman 1999) |
| `acceptance` | Nokta ve sıkı kabul kararı |
| `proportional_bias` | Fark ~ ortalama regresyonu; eğim anlamlıysa orantısal hata var |
| `icc_a1` | ICC(A,1), mutlak uyum |
| `categorical` | Klinik eşikte duyarlılık, özgüllük ve kappa |
| `simulate_power` | Örneklem büyüklüğü planlama simülasyonu |

````python
# ---------------------------------------------------------------------------
# İSTATİSTİK FONKSİYONLARI
# ---------------------------------------------------------------------------
def differences(new, ref, scale):
    """Fark: 'abs' = new − ref ; 'pct' = (new − ref) / ortalama × 100"""
    new, ref = np.asarray(new, float), np.asarray(ref, float)
    if scale == "abs":
        return new - ref
    return (new - ref) / ((new + ref) / 2) * 100


def bland_altman(diff, alpha=0.05):
    """Bias, SD, %95 LoA ve güven aralıkları (Bland & Altman 1999, SMMR).
    LoA standart hatası ≈ SD × sqrt(1/n + z²/(2(n−1)))"""
    d = np.asarray(diff, float)
    d = d[~np.isnan(d)]
    n = len(d)
    bias, sd = d.mean(), d.std(ddof=1)
    z = stats.norm.ppf(0.975)
    t = stats.t.ppf(1 - alpha / 2, n - 1)
    se_bias = sd / np.sqrt(n)
    se_loa = sd * np.sqrt(1 / n + z ** 2 / (2 * (n - 1)))
    lo, hi = bias - z * sd, bias + z * sd
    return {
        "n": n, "bias": bias, "sd": sd,
        "bias_ci_lo": bias - t * se_bias, "bias_ci_hi": bias + t * se_bias,
        "loa_lo": lo, "loa_hi": hi,
        "loa_lo_ci_lo": lo - t * se_loa, "loa_lo_ci_hi": lo + t * se_loa,
        "loa_hi_ci_lo": hi - t * se_loa, "loa_hi_ci_hi": hi + t * se_loa,
    }


def acceptance(res, limit):
    """İki seviye karar:
       point  : LoA nokta tahmini ±limit içinde
       strict : LoA'nın %95 GA'sının dış ucu da ±limit içinde (daha sıkı)"""
    if limit is None:
        return {"point": None, "strict": None}
    point = (res["loa_lo"] >= -limit) and (res["loa_hi"] <= limit)
    strict = (res["loa_lo_ci_lo"] >= -limit) and (res["loa_hi_ci_hi"] <= limit)
    return {"point": point, "strict": strict}


def proportional_bias(new, ref, diff):
    """Farkın büyüklükle ilişkisi: fark ~ ortalama regresyonu.
    Eğim anlamlıysa orantısal hata var -> yüzde fark veya log dönüşümü düşünülmeli."""
    mean = (np.asarray(new) + np.asarray(ref)) / 2
    r = stats.linregress(mean, diff)
    return {"slope": r.slope, "intercept": r.intercept, "p": r.pvalue, "r": r.rvalue}


def icc_a1(x, y):
    """ICC(A,1): iki yönlü rastgele etkiler, mutlak uyum, tek ölçüm (McGraw & Wong 1996)."""
    data = np.column_stack([x, y])
    n, k = data.shape
    gm = data.mean()
    ms_r = k * ((data.mean(axis=1) - gm) ** 2).sum() / (n - 1)
    ms_c = n * ((data.mean(axis=0) - gm) ** 2).sum() / (k - 1)
    ss_e = ((data - data.mean(axis=1, keepdims=True) - data.mean(axis=0, keepdims=True) + gm) ** 2).sum()
    ms_e = ss_e / ((n - 1) * (k - 1))
    return (ms_r - ms_e) / (ms_r + (k - 1) * ms_e + k * (ms_c - ms_e) / n)


def wilson(k, n, z=1.96):
    if n == 0:
        return np.nan, np.nan, np.nan
    p = k / n
    den = 1 + z ** 2 / n
    c = (p + z ** 2 / (2 * n)) / den
    h = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2)) / den
    return p, c - h, c + h


def categorical(ai, ref, cutoff):
    """Klinik eşik değerde sınıflama uyumu (referans pozitif = ref > cutoff)."""
    a, r = np.asarray(ai) > cutoff, np.asarray(ref) > cutoff
    tp, tn = int((a & r).sum()), int((~a & ~r).sum())
    fp, fn = int((a & ~r).sum()), int((~a & r).sum())
    n = tp + tn + fp + fn
    po = (tp + tn) / n
    pe = ((tp + fp) * (tp + fn) + (fn + tn) * (fp + tn)) / n ** 2
    kappa = (po - pe) / (1 - pe) if pe < 1 else np.nan
    return {"tp": tp, "tn": tn, "fp": fp, "fn": fn,
            "sens": wilson(tp, tp + fn), "spec": wilson(tn, tn + fp),
            "agreement": po, "kappa": kappa}


def simulate_power(bias, sd, delta, n, nsim=2000, rng=None):
    """Planlama: n vakada LoA %95 GA'sının ±delta içinde kalma olasılığı (strict kriter)."""
    rng = rng or np.random.default_rng(1)
    ok = 0
    for _ in range(nsim):
        res = bland_altman(rng.normal(bias, sd, n))
        ok += (res["loa_lo_ci_lo"] >= -delta) and (res["loa_hi_ci_hi"] <= delta)
    return ok / nsim
````

## 3. Grafik fonksiyonları
Grafikler hem notebook içinde gösterilir hem de çıktı klasörüne PNG olarak kaydedilir.

````python
# ---------------------------------------------------------------------------
# GRAFİKLER
# ---------------------------------------------------------------------------
def style_axes(ax):
    for s in ["top", "right"]:
        ax.spines[s].set_visible(False)
    ax.grid(alpha=0.25)


def plot_ba(new, ref, scale, res, limit, title, ylabel, xlabel, path, highlight=None):
    mean = (np.asarray(new) + np.asarray(ref)) / 2
    diff = differences(new, ref, scale)
    fig, ax = plt.subplots(figsize=(10, 6.5))
    if limit is not None:
        ax.axhspan(-limit, limit, color=COL["band"], zorder=0,
                   label=f"Kabul sınırı ±{limit:g}")
    ax.scatter(mean, diff, s=28, color=COL["ink"], alpha=0.75, zorder=3)
    if highlight is not None and highlight.any():
        ax.scatter(mean[highlight], diff[highlight], s=90, facecolors="none",
                   edgecolors=COL["fail"], linewidths=1.6, zorder=4, label="LoA dışı vaka")
    xr = [mean.min(), mean.max()]
    ax.axhline(0, color=COL["grey"], lw=1, ls=":")
    ax.axhline(res["bias"], color=COL["blue"], lw=2.2, label=f"Bias {res['bias']:+.2f}")
    ax.axhspan(res["bias_ci_lo"], res["bias_ci_hi"], color=COL["blue"], alpha=0.12)
    for key, lab in [("loa_lo", "Alt LoA"), ("loa_hi", "Üst LoA")]:
        ax.axhline(res[key], color=COL["accent"], lw=2, ls="--",
                   label=f"{lab} {res[key]:+.2f}")
        ax.axhspan(res[key + "_ci_lo"], res[key + "_ci_hi"], color=COL["accent"], alpha=0.12)
    pb = proportional_bias(new, ref, diff)
    xs = np.linspace(*xr, 50)
    ax.plot(xs, pb["intercept"] + pb["slope"] * xs, color=COL["ink"], lw=1, ls="-.",
            alpha=0.6, label=f"Trend (eğim p={pb['p']:.3f})")
    ax.set_title(title, fontsize=13, loc="left", color=COL["ink"])
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    style_axes(ax)
    ax.legend(fontsize=8.5, loc="upper left", bbox_to_anchor=(1.01, 1), frameon=False)
    fig.tight_layout()
    fig.savefig(path, dpi=160)
    plt.show()


def plot_scatter(ai, ref, unit, cutoff, title, path):
    fig, ax = plt.subplots(figsize=(7, 7))
    lo, hi = min(ai.min(), ref.min()), max(ai.max(), ref.max())
    ax.plot([lo, hi], [lo, hi], color=COL["grey"], lw=1, ls="--", label="Özdeşlik (y = x)")
    ax.scatter(ref, ai, s=26, color=COL["ink"], alpha=0.75)
    ax.axvline(cutoff, color=COL["accent"], lw=1, ls=":")
    ax.axhline(cutoff, color=COL["accent"], lw=1, ls=":", label=f"Klinik eşik {cutoff:g} {unit}")
    r = stats.pearsonr(ref, ai)[0]
    ax.set_title(f"{title}\nPearson r = {r:.3f}  (uyum için tek başına yeterli değil)",
                 fontsize=11, loc="left")
    ax.set_xlabel(f"Referans: 3 uzman ortalaması ({unit})")
    ax.set_ylabel(f"AI ({unit})")
    ax.set_aspect("equal")
    style_axes(ax)
    ax.legend(fontsize=9, frameon=False)
    fig.tight_layout()
    fig.savefig(path, dpi=160)
    plt.show()


def plot_interval_rows(rows, limit, title, xlabel, path, label_col="label"):
    """Her satır: bias noktası + LoA aralığı (forest tarzı)."""
    fig, ax = plt.subplots(figsize=(10, 0.55 * len(rows) + 1.8))
    y = np.arange(len(rows))[::-1]
    if limit is not None:
        ax.axvspan(-limit, limit, color=COL["band"], zorder=0, label=f"Kabul ±{limit:g}")
    for yi, (_, r) in zip(y, rows.iterrows()):
        c = COL["accent"] if r.get("is_ai", False) else COL["ink"]
        ax.plot([r["loa_lo"], r["loa_hi"]], [yi, yi], color=c, lw=3, solid_capstyle="butt")
        ax.plot(r["bias"], yi, "o", color=COL["blue"], ms=7, zorder=3)
        ax.text(r["loa_hi"], yi + 0.18, f" n={int(r['n'])}", fontsize=8, color=COL["grey"])
    ax.axvline(0, color=COL["grey"], lw=1, ls=":")
    ax.set_yticks(y)
    ax.set_yticklabels(rows[label_col], fontsize=9)
    ax.set_xlabel(xlabel)
    ax.set_title(title, fontsize=12, loc="left")
    style_axes(ax)
    if limit is not None:
        ax.legend(fontsize=9, frameon=False, loc="lower right")
    fig.tight_layout()
    fig.savefig(path, dpi=160)
    plt.show()


def plot_power(cfg_p, path):
    ns = np.arange(20, 401, 20)
    rng = np.random.default_rng(7)
    pw = [simulate_power(cfg_p["bias"], cfg_p["sd"], cfg_p["delta"], n, 1500, rng) for n in ns]
    fig, ax = plt.subplots(figsize=(9, 5.5))
    ax.plot(ns, pw, "o-", color=COL["ink"])
    ax.axhline(0.8, color=COL["accent"], ls="--", lw=1.2, label="%80 güç")
    ax.axhline(0.9, color=COL["blue"], ls="--", lw=1.2, label="%90 güç")
    ax.set_ylim(0, 1.02)
    ax.set_xlabel("Analiz edilebilir vaka sayısı (n)")
    ax.set_ylabel("P(LoA %95 GA ±kabul sınırı içinde)")
    ax.set_title(f"Örneklem planlama (simülasyon): bias={cfg_p['bias']}, SD={cfg_p['sd']}, "
                 f"sınır=±{cfg_p['delta']}", fontsize=11, loc="left")
    style_axes(ax)
    ax.legend(frameon=False)
    fig.tight_layout()
    fig.savefig(path, dpi=160)
    plt.show()
    return dict(zip(ns.tolist(), pw))
````

## 4. Ana analiz fonksiyonu
`run_study()` tüm adımları sırayla yapar: hasta akışı, birincil ve ikincil sonlanım, ek ölçüler, klinik eşik, insan okuyucularla karşılaştırma, alt gruplar, grafikler ve (isteğe bağlı) örneklem planlama.

````python
# ---------------------------------------------------------------------------
# ANA AKIŞ
# ---------------------------------------------------------------------------
def fmt_res(res, unit=""):
    return (f"n={res['n']}  bias={res['bias']:+.2f}{unit} "
            f"(95% GA {res['bias_ci_lo']:+.2f} / {res['bias_ci_hi']:+.2f})  SD={res['sd']:.2f}\n"
            f"      LoA = {res['loa_lo']:+.2f} / {res['loa_hi']:+.2f}{unit}   "
            f"[alt LoA GA {res['loa_lo_ci_lo']:+.2f} / {res['loa_lo_ci_hi']:+.2f};  "
            f"üst LoA GA {res['loa_hi_ci_lo']:+.2f} / {res['loa_hi_ci_hi']:+.2f}]")


def verdict(acc):
    if acc["point"] is None:
        return "Önceden sınır tanımlanmadı (tanımlayıcı/keşifsel)"
    s = lambda v: "PASS" if v else "FAIL"
    return f"LoA nokta tahmini: {s(acc['point'])}   |   LoA %95 GA (sıkı): {s(acc['strict'])}"


def run_study(csv_path, study=None, out=None, run_power=True):
    """Bir çalışmanın tüm analizini yapar, grafikleri gösterir ve dosyaya kaydeder.
    csv_path : CSV dosyasının yolu
    study    : "bladder" | "liver" | None (None ise sütunlardan anlaşılır)
    out      : çıktı klasörü (varsayılan results_<study>)
    run_power: örneklem planlama simülasyonu (birkaç dakika sürebilir)
    Döndürür : sonuç tablolarını içeren sözlük"""
    raw = pd.read_csv(csv_path)
    study = study or ("bladder" if "ai_ml" in raw.columns else "liver" if "ai_cm" in raw.columns else None)
    if study is None:
        raise ValueError("Çalışma tipi anlaşılamadı; study='bladder' veya 'liver' verin.")
    cfg = CONFIG[study]
    out = out or f"results_{study}"
    os.makedirs(out, exist_ok=True)
    u = cfg["unit"]
    lines = []
    P = lines.append

    # --- Akış (CONSORT benzeri) ---
    n_all = len(raw)
    excl = raw[raw["analyzable"] != "Yes"]
    df = raw[raw["analyzable"] == "Yes"].copy()
    df["reference"] = df[cfg["readers"]].mean(axis=1)
    df["age_group"] = pd.cut(df["age"], [0, 39, 64, 200], labels=["18-39", "40-64", "65+"]).astype(str)
    ai, ref = df[cfg["ai"]].to_numpy(), df["reference"].to_numpy()

    P("=" * 78)
    P(f"{cfg['title']} — Standalone performans analizi (SENTETİK EĞİTİM VERİSİ)")
    P("=" * 78)
    P(f"Kayıtlı vaka: {n_all}   Dışlanan: {len(excl)}   Analiz edilen: {len(df)}")
    for reason, k in excl["exclusion_reason"].value_counts().items():
        P(f"   - {reason}: {k}")
    P(f"Referans: {len(cfg['readers'])} bağımsız, kör uzman okuyucunun ortalaması")
    P(f"Referans aralığı: {ref.min():.1f} – {ref.max():.1f} {u} (medyan {np.median(ref):.1f})")

    endpoints = []

    # --- Birincil sonlanım ---
    pr = cfg["primary"]
    d1 = differences(ai, ref, pr["scale"])
    r1 = bland_altman(d1)
    a1 = acceptance(r1, pr["limit"])
    pb1 = proportional_bias(ai, ref, d1)
    sw = stats.shapiro(d1)
    unit1 = "%" if pr["scale"] == "pct" else f" {u}"
    P("\n[BİRİNCİL SONLANIM] " + pr["name"] + f"   kabul sınırı ±{pr['limit']:g}{unit1}")
    P("   " + fmt_res(r1))
    P("   KARAR: " + verdict(a1))
    P(f"   Orantısal hata testi: eğim={pb1['slope']:+.4f}, p={pb1['p']:.4f}"
      + ("  -> fark büyüklükle değişiyor, yorumlarken dikkat" if pb1["p"] < 0.05 else "  -> anlamlı değil"))
    P(f"   Farkların normalliği (Shapiro-Wilk): W={sw.statistic:.3f}, p={sw.pvalue:.4f}")
    if sw.pvalue < 0.05:
        q = np.percentile(d1, [2.5, 97.5])
        P(f"   Normallik zayıf -> nonparametrik LoA (2.5–97.5. persentil): {q[0]:+.2f} / {q[1]:+.2f}")
    outside = (d1 < r1["loa_lo"]) | (d1 > r1["loa_hi"])
    P(f"   LoA dışında kalan vaka: {outside.sum()} / {len(d1)} ({outside.mean()*100:.1f}%; beklenen ≈ %5)")
    endpoints.append({"endpoint": "primary: " + pr["name"], "limit": pr["limit"], **r1,
                      "pass_point": a1["point"], "pass_strict": a1["strict"],
                      "prop_bias_slope": pb1["slope"], "prop_bias_p": pb1["p"]})

    # --- İkincil sonlanım ---
    se = cfg["secondary"]
    mask = np.ones(len(df), bool) if not se.get("min_ref") else ref >= se["min_ref"]
    d2 = differences(ai[mask], ref[mask], se["scale"])
    r2 = bland_altman(d2)
    a2 = acceptance(r2, se["limit"])
    P("\n[İKİNCİL SONLANIM] " + se["name"])
    P("   " + fmt_res(r2))
    P("   KARAR: " + verdict(a2))
    endpoints.append({"endpoint": "secondary: " + se["name"], "limit": se["limit"], **r2,
                      "pass_point": a2["point"], "pass_strict": a2["strict"]})

    # --- Ek uyum ölçüleri ---
    icc = icc_a1(ai, ref)
    rp = stats.pearsonr(ai, ref)[0]
    mae = np.mean(np.abs(ai - ref))
    P(f"\n[EK ÖLÇÜLER] ICC(A,1) = {icc:.3f}   Pearson r = {rp:.3f}   MAE = {mae:.2f} {u}")
    P("   Not: r ve ICC yüksek olsa da karar LoA ile kabul sınırına göre verilir.")

    # --- Klinik eşikte sınıflama ---
    ct = cfg["cutoff"]
    cat = categorical(ai, ref, ct["value"])
    P(f"\n[KLİNİK EŞİK] {ct['label']}")
    P(f"   TP={cat['tp']} FN={cat['fn']} FP={cat['fp']} TN={cat['tn']}")
    P(f"   Duyarlılık = {cat['sens'][0]:.3f} (Wilson %95 GA {cat['sens'][1]:.3f}–{cat['sens'][2]:.3f})")
    P(f"   Özgüllük   = {cat['spec'][0]:.3f} (Wilson %95 GA {cat['spec'][1]:.3f}–{cat['spec'][2]:.3f})")
    P(f"   Toplam uyum = {cat['agreement']:.3f}   Cohen kappa = {cat['kappa']:.3f}")

    # --- İnsan okuyucular arası karşılaştırma ---
    bench = []
    rd = cfg["readers"]
    for i in range(len(rd)):
        for j in range(i + 1, len(rd)):
            rr = bland_altman(differences(df[rd[i]], df[rd[j]], pr["scale"]))
            bench.append({"label": f"Okuyucu {i+1} − Okuyucu {j+1}", **rr, "is_ai": False})
    for i in range(len(rd)):
        others = df[[c for k, c in enumerate(rd) if k != i]].mean(axis=1)
        rr = bland_altman(differences(df[rd[i]], others, pr["scale"]))
        bench.append({"label": f"Okuyucu {i+1} − diğer 2'nin ort.", **rr, "is_ai": False})
    bench.append({"label": "AI − 3 okuyucu ort.", **r1, "is_ai": True})
    bench = pd.DataFrame(bench)
    P("\n[İNSAN KARŞILAŞTIRMASI] (birincil ölçekte)")
    for _, b in bench.iterrows():
        P(f"   {b['label']:<30} bias={b['bias']:+7.2f}   LoA {b['loa_lo']:+7.2f} / {b['loa_hi']:+7.2f}")
    P("   Yorum: AI'ın LoA genişliği, okuyucuların birbirine göre LoA genişliğiyle kıyaslanır.")

    # --- Alt gruplar ---
    sub_rows = []
    for col in SUBGROUP_COLS:
        if col not in df.columns:
            continue
        for lvl, g in df.groupby(col):
            if len(g) < 5:
                continue
            gd = differences(g[cfg["ai"]], g["reference"], pr["scale"])
            gr = bland_altman(gd)
            sub_rows.append({"label": f"{col}: {lvl}", "subgroup": col, "level": lvl, **gr})
    subs = pd.DataFrame(sub_rows)
    P("\n[ALT GRUPLAR] (keşifsel; birincil ölçekte)")
    for _, s in subs.iterrows():
        flag = ""
        if pr["limit"] is not None and (s["loa_lo"] < -pr["limit"] or s["loa_hi"] > pr["limit"]):
            flag = "  <- sınırı aşıyor"
        P(f"   {s['label']:<30} n={int(s['n']):>3}  bias={s['bias']:+7.2f}  "
          f"LoA {s['loa_lo']:+7.2f} / {s['loa_hi']:+7.2f}{flag}")

    # --- Grafikler ---
    ylabel1 = (f"Fark: (AI − Referans) / ortalama × 100 (%)" if pr["scale"] == "pct"
               else f"Fark: AI − Referans ({u})")
    plot_ba(ai, ref, pr["scale"], r1, pr["limit"],
            f"{cfg['title']} — Birincil Bland-Altman (n={r1['n']})", ylabel1,
            f"Ortalama: (AI + Referans) / 2 ({u})",
            os.path.join(out, "01_bland_altman_primary.png"), highlight=outside)
    plot_scatter(ai, ref, u, ct["value"], cfg["title"],
                 os.path.join(out, "02_scatter_ai_vs_reference.png"))
    ylabel2 = (f"Fark: (AI − Referans) / ortalama × 100 (%)" if se["scale"] == "pct"
               else f"Fark: AI − Referans ({u})")
    plot_ba(ai[mask], ref[mask], se["scale"], r2, se["limit"],
            f"{cfg['title']} — İkincil: {se['name']} (n={r2['n']})", ylabel2,
            f"Ortalama: (AI + Referans) / 2 ({u})",
            os.path.join(out, "03_bland_altman_secondary.png"))
    plot_interval_rows(bench, pr["limit"],
                       "AI ve insan okuyucular: bias (nokta) ve %95 LoA (çizgi)",
                       ylabel1, os.path.join(out, "04_ai_vs_human_loa.png"))
    if len(subs):
        plot_interval_rows(subs, pr["limit"], "Alt gruplarda bias ve %95 LoA (keşifsel)",
                           ylabel1, os.path.join(out, "05_subgroup_forest.png"))

    if run_power:
        pw = plot_power(cfg["power"], os.path.join(out, "06_sample_size_power.png"))
        P("\n[ÖRNEKLEM PLANLAMA] (simülasyon; strict kriter)")
        P(f"   Varsayım: bias={cfg['power']['bias']}, SD={cfg['power']['sd']}, sınır=±{cfg['power']['delta']}")
        for target in (0.8, 0.9):
            n_ok = next((n for n, p in pw.items() if p >= target), None)
            P(f"   %{int(target*100)} güç için gereken en küçük n (20'lik adımlarla): "
              + (str(n_ok) if n_ok else "400'e kadar ulaşılamadı (SD varsayımını veya sınırı gözden geçirin)"))
        P("   Not: gerçek planlamada SD pilot veriden alınır ve %10–20 dışlanma payı eklenir.")

    # --- Dosyalar ---
    pd.DataFrame(endpoints).round(4).to_csv(os.path.join(out, "endpoints.csv"), index=False)
    bench.round(4).to_csv(os.path.join(out, "reader_benchmark.csv"), index=False)
    subs.round(4).to_csv(os.path.join(out, "subgroups.csv"), index=False)
    report = "\n".join(lines)
    with open(os.path.join(out, "summary_report.txt"), "w", encoding="utf-8") as f:
        f.write(report + "\n")
    print(report)
    print(f"\nÇıktılar: {os.path.abspath(out)}")
    return {"data": df, "endpoints": pd.DataFrame(endpoints), "reader_benchmark": bench,
            "subgroups": subs, "report": report, "out_dir": os.path.abspath(out)}
````
