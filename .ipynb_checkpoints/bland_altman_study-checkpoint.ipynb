### Synthetic data generator / Sentetik veri üretici

**EN:** Runs only if the CSV is missing. It simulates 3 sites, cart-based and handheld devices, 3 readers with small systematic differences (reader 3 measures slightly larger), and a few AI outliers.

**TR:** Yalnızca CSV yoksa çalışır. 3 merkez, cart-based ve handheld cihazlar, küçük sistematik farkları olan 3 okuyucu (okuyucu 3 biraz büyük ölçüyor) ve birkaç AI aykırı değeri simüle eder.

````python
SEED = 2026

def _demographics(rng, n):
    age = np.clip(rng.normal(55, 17, n), 18, 92).round().astype(int)
    sex = rng.choice(["F", "M"], n, p=[0.52, 0.48])
    bmi = np.clip(rng.lognormal(np.log(27), 0.18, n), 17, 48).round(1)
    bmi_cat = pd.cut(bmi, [0, 25, 30, 100], labels=["<25", "25-30", ">=30"]).astype(str)
    site = rng.choice(["Site_A", "Site_B", "Site_C"], n, p=[0.40, 0.35, 0.25])
    device = rng.choice(["Cart-based", "Handheld"], n, p=[0.6, 0.4])
    return age, sex, bmi, bmi_cat, site, device

def make_bladder(rng, n=150):
    age, sex, bmi, bmi_cat, site, device = _demographics(rng, n)
    true_v = np.clip(rng.lognormal(np.log(170), 0.75, n), 15, 650)   # true volume / gerçek hacim
    obese, handheld = bmi >= 30, device == "Handheld"
    readers = {}
    for name, b in {"reader1_ml": -0.02, "reader2_ml": 0.00, "reader3_ml": 0.025}.items():
        sd = (3 + 0.07 * true_v) * np.where(obese, 1.15, 1.0)
        readers[name] = np.clip(true_v * (1 + b) + rng.normal(0, sd), 5, None)
    ai_sd = (2.5 + 0.045 * true_v) * np.where(obese, 1.25, 1.0) * np.where(handheld, 1.15, 1.0)
    ai = true_v * 1.015 + 2 + rng.normal(0, ai_sd)
    out_idx = rng.choice(n, 2, replace=False)
    ai[out_idx] += np.array([70, -55])                                # segmentation errors / segmentasyon hataları
    ai = np.clip(ai, 5, None)
    quality = rng.choice(["Good", "Adequate", "Poor"], n, p=[0.62, 0.32, 0.06])
    excl = np.full(n, "", dtype=object)
    excl[quality == "Poor"] = "Poor image quality"
    excl[rng.choice(np.where(quality != "Poor")[0], 3, replace=False)] = "Bladder not fully in field of view"
    df = pd.DataFrame({"case_id": [f"BLD-{i+1:03d}" for i in range(n)], "site": site,
                       "device_type": device, "age": age, "sex": sex, "bmi": bmi,
                       "bmi_category": bmi_cat, "image_quality": quality,
                       "analyzable": np.where(excl == "", "Yes", "No"), "exclusion_reason": excl,
                       **{k: v.round(1) for k, v in readers.items()}, "ai_ml": ai.round(1)})
    df.loc[df.analyzable == "No", list(readers) + ["ai_ml"]] = np.nan
    return df

def make_liver(rng, n=160):
    age, sex, bmi, bmi_cat, site, device = _demographics(rng, n)
    true_l = np.clip(rng.normal(15.3, 2.3, n) + np.where(bmi >= 30, 0.8, 0), 10, 23)
    obese = bmi >= 30
    readers = {}
    for name, b in {"reader1_cm": -0.02, "reader2_cm": 0.0, "reader3_cm": 0.02}.items():
        readers[name] = true_l * (1 + b) + rng.normal(0, 0.052 * true_l * np.where(obese, 1.2, 1.0))
    ai = true_l * 0.985 + rng.normal(0, 0.038 * true_l * np.where(obese, 1.3, 1.0))
    ai[rng.choice(n, 1)] -= 3.0                                       # missed Riedel lobe / kaçırılan Riedel lobu
    quality = rng.choice(["Good", "Adequate", "Poor"], n, p=[0.58, 0.35, 0.07])
    excl = np.full(n, "", dtype=object)
    excl[quality == "Poor"] = "Poor image quality / rib shadowing"
    excl[rng.choice(np.where(quality != "Poor")[0], 3, replace=False)] = "Liver tip not visualized"
    df = pd.DataFrame({"case_id": [f"LVR-{i+1:03d}" for i in range(n)], "site": site,
                       "device_type": device, "age": age, "sex": sex, "bmi": bmi,
                       "bmi_category": bmi_cat, "image_quality": quality,
                       "analyzable": np.where(excl == "", "Yes", "No"), "exclusion_reason": excl,
                       **{k: v.round(2) for k, v in readers.items()}, "ai_cm": ai.round(2)})
    df.loc[df.analyzable == "No", list(readers) + ["ai_cm"]] = np.nan
    return df

def ensure_csvs():
    # Same RNG order as the original script -> identical CSVs / Orijinal betikle aynı sıra -> aynı CSV'ler
    if not (os.path.exists("bladder_study.csv") and os.path.exists("liver_study.csv")):
        rng = np.random.default_rng(SEED)
        make_bladder(rng).to_csv("bladder_study.csv", index=False)
        make_liver(rng).to_csv("liver_study.csv", index=False)
        print("CSV files generated / CSV dosyaları üretildi")

if CSV_PATH is None:
    ensure_csvs()
    CSV_PATH = f"{STUDY}_study.csv"
print("Data file / Veri dosyası:", CSV_PATH)
````

### Statistical functions / İstatistik fonksiyonları

| Function / Fonksiyon | EN | TR |
|---|---|---|
| `differences` | `abs`: AI − ref; `pct`: (AI − ref) / mean × 100 | `abs`: AI − ref; `pct`: (AI − ref) / ortalama × 100 |
| `bland_altman` | Bias, SD, 95 % LoA and their CIs (Bland & Altman 1999) | Bias, SD, %95 LoA ve güven aralıkları (Bland & Altman 1999) |
| `acceptance` | *point*: LoA inside the limit; *strict*: LoA CIs also inside | *nokta*: LoA sınır içinde; *sıkı*: LoA güven aralıkları da sınır içinde |
| `proportional_bias` | Regression of difference on mean | Farkın ortalamaya göre regresyonu |
| `icc_a1` | ICC(A,1), absolute agreement | ICC(A,1), mutlak uyum |
| `categorical` | Sensitivity, specificity, kappa at a clinical cut-off | Klinik eşikte duyarlılık, özgüllük, kappa |

````python
def differences(new, ref, scale):
    # 'abs' = new − ref ; 'pct' = (new − ref) / mean × 100
    new, ref = np.asarray(new, float), np.asarray(ref, float)
    return new - ref if scale == "abs" else (new - ref) / ((new + ref) / 2) * 100

def bland_altman(diff, alpha=0.05):
    # SE(LoA) ≈ SD × sqrt(1/n + z²/(2(n−1)))  — Bland & Altman 1999
    d = np.asarray(diff, float); d = d[~np.isnan(d)]
    n, bias, sd = len(d), d.mean(), d.std(ddof=1)
    z, t = stats.norm.ppf(0.975), stats.t.ppf(1 - alpha / 2, n - 1)
    se_b, se_l = sd / np.sqrt(n), sd * np.sqrt(1 / n + z**2 / (2 * (n - 1)))
    lo, hi = bias - z * sd, bias + z * sd
    return {"n": n, "bias": bias, "sd": sd,
            "bias_ci_lo": bias - t * se_b, "bias_ci_hi": bias + t * se_b,
            "loa_lo": lo, "loa_hi": hi,
            "loa_lo_ci_lo": lo - t * se_l, "loa_lo_ci_hi": lo + t * se_l,
            "loa_hi_ci_lo": hi - t * se_l, "loa_hi_ci_hi": hi + t * se_l}

def acceptance(res, limit):
    if limit is None:
        return {"point": None, "strict": None}
    return {"point": res["loa_lo"] >= -limit and res["loa_hi"] <= limit,
            "strict": res["loa_lo_ci_lo"] >= -limit and res["loa_hi_ci_hi"] <= limit}

def verdict(acc):
    if acc["point"] is None:
        return "No pre-specified limit (descriptive) / Önceden sınır yok (tanımlayıcı)"
    s = lambda v: "PASS ✅" if v else "FAIL ❌"
    return (f"LoA point estimate / LoA nokta tahmini: {s(acc['point'])}   |   "
            f"LoA 95% CI (strict) / LoA %95 GA (sıkı): {s(acc['strict'])}")

def proportional_bias(new, ref, diff):
    mean = (np.asarray(new) + np.asarray(ref)) / 2
    r = stats.linregress(mean, diff)
    return {"slope": r.slope, "intercept": r.intercept, "p": r.pvalue}

def icc_a1(x, y):
    # ICC(A,1): two-way random, absolute agreement, single measure (McGraw & Wong 1996)
    data = np.column_stack([x, y]); n, k = data.shape; gm = data.mean()
    ms_r = k * ((data.mean(1) - gm) ** 2).sum() / (n - 1)
    ms_c = n * ((data.mean(0) - gm) ** 2).sum() / (k - 1)
    ss_e = ((data - data.mean(1, keepdims=True) - data.mean(0, keepdims=True) + gm) ** 2).sum()
    ms_e = ss_e / ((n - 1) * (k - 1))
    return (ms_r - ms_e) / (ms_r + (k - 1) * ms_e + k * (ms_c - ms_e) / n)

def wilson(k, n, z=1.96):
    p = k / n; den = 1 + z**2 / n
    c = (p + z**2 / (2 * n)) / den; h = z * np.sqrt(p * (1 - p) / n + z**2 / (4 * n**2)) / den
    return p, c - h, c + h

def categorical(ai, ref, cutoff):
    a, r = np.asarray(ai) > cutoff, np.asarray(ref) > cutoff
    tp, tn, fp, fn = int((a & r).sum()), int((~a & ~r).sum()), int((a & ~r).sum()), int((~a & r).sum())
    n = tp + tn + fp + fn; po = (tp + tn) / n
    pe = ((tp + fp) * (tp + fn) + (fn + tn) * (fp + tn)) / n**2
    return {"tp": tp, "tn": tn, "fp": fp, "fn": fn, "sens": wilson(tp, tp + fn),
            "spec": wilson(tn, tn + fp), "agreement": po, "kappa": (po - pe) / (1 - pe)}

def res_table(res):
    # One-row summary table / Tek satırlık özet tablo
    return pd.DataFrame([{
        "n": res["n"], "Bias": res["bias"],
        "Bias 95% CI / GA": f"{res['bias_ci_lo']:+.2f} / {res['bias_ci_hi']:+.2f}",
        "SD": res["sd"], "Lower LoA / Alt LoA": res["loa_lo"],
        "Lower LoA CI / Alt LoA GA": f"{res['loa_lo_ci_lo']:+.2f} / {res['loa_lo_ci_hi']:+.2f}",
        "Upper LoA / Üst LoA": res["loa_hi"],
        "Upper LoA CI / Üst LoA GA": f"{res['loa_hi_ci_lo']:+.2f} / {res['loa_hi_ci_hi']:+.2f}"}])
````

### Plot functions / Grafik fonksiyonları

````python
def save(fig, name):
    if SAVE_OUTPUTS:
        fig.savefig(os.path.join(OUT, name), dpi=160, bbox_inches="tight")

def plot_ba(new, ref, scale, res, limit, title, ylabel, fname, highlight=None):
    mean = (np.asarray(new) + np.asarray(ref)) / 2
    diff = differences(new, ref, scale)
    fig, ax = plt.subplots(figsize=(11, 6.5))
    if limit is not None:
        ax.axhspan(-limit, limit, color=COL["band"], zorder=0,
                   label=f"Acceptance limit / Kabul sınırı ±{limit:g}")
    ax.scatter(mean, diff, s=28, color=COL["ink"], alpha=0.75, zorder=3)
    if highlight is not None and highlight.any():
        ax.scatter(mean[highlight], diff[highlight], s=90, facecolors="none", edgecolors=COL["fail"],
                   linewidths=1.6, zorder=4, label="Outside LoA / LoA dışı")
    ax.axhline(0, color=COL["grey"], lw=1, ls=":")
    ax.axhline(res["bias"], color=COL["blue"], lw=2.2, label=f"Bias {res['bias']:+.2f}")
    ax.axhspan(res["bias_ci_lo"], res["bias_ci_hi"], color=COL["blue"], alpha=0.12)
    for key, lab in [("loa_lo", "Lower LoA / Alt LoA"), ("loa_hi", "Upper LoA / Üst LoA")]:
        ax.axhline(res[key], color=COL["accent"], lw=2, ls="--", label=f"{lab} {res[key]:+.2f}")
        ax.axhspan(res[key + "_ci_lo"], res[key + "_ci_hi"], color=COL["accent"], alpha=0.12)
    pb = proportional_bias(new, ref, diff)
    xs = np.linspace(mean.min(), mean.max(), 50)
    ax.plot(xs, pb["intercept"] + pb["slope"] * xs, color=COL["ink"], lw=1, ls="-.", alpha=0.6,
            label=f"Trend (slope/eğim p={pb['p']:.3f})")
    ax.set_title(title, loc="left", fontsize=12, color=COL["ink"])
    ax.set_xlabel(f"Mean / Ortalama: (AI + Reference/Referans) / 2 ({U})")
    ax.set_ylabel(ylabel)
    ax.grid(alpha=0.25)
    ax.legend(fontsize=8.5, loc="upper left", bbox_to_anchor=(1.01, 1), frameon=False)
    fig.tight_layout(); save(fig, fname); plt.show()

def plot_intervals(rows, limit, title, xlabel, fname):
    fig, ax = plt.subplots(figsize=(10, 0.5 * len(rows) + 1.8))
    y = np.arange(len(rows))[::-1]
    if limit is not None:
        ax.axvspan(-limit, limit, color=COL["band"], zorder=0, label=f"Limit / Sınır ±{limit:g}")
    for yi, (_, r) in zip(y, rows.iterrows()):
        c = COL["accent"] if r.get("is_ai", False) else COL["ink"]
        ax.plot([r["loa_lo"], r["loa_hi"]], [yi, yi], color=c, lw=3, solid_capstyle="butt")
        ax.plot(r["bias"], yi, "o", color=COL["blue"], ms=7, zorder=3)
        ax.text(r["loa_hi"], yi + 0.18, f" n={int(r['n'])}", fontsize=8, color=COL["grey"])
    ax.axvline(0, color=COL["grey"], lw=1, ls=":")
    ax.set_yticks(y); ax.set_yticklabels(rows["label"], fontsize=9)
    ax.set_xlabel(xlabel); ax.set_title(title, loc="left", fontsize=12); ax.grid(alpha=0.25)
    if limit is not None:
        ax.legend(fontsize=9, frameon=False, loc="upper left", bbox_to_anchor=(1.01, 1))
    fig.tight_layout(); save(fig, fname); plt.show()

def ylabel_for(scale):
    return ("Difference / Fark: (AI − Ref) / mean × 100 (%)" if scale == "pct"
            else f"Difference / Fark: AI − Reference/Referans ({U})")
````

## 1. Data loading and patient flow / Veri yükleme ve hasta akışı

**EN:** Non-analyzable cases are excluded and reported, as in a CONSORT/STARD flow. The reference standard is the mean of the 3 readers.

**TR:** Analiz edilemeyen vakalar dışlanır ve CONSORT/STARD akışındaki gibi raporlanır. Referans standart 3 okuyucunun ortalamasıdır.

````python
raw = pd.read_csv(CSV_PATH)
excl = raw[raw["analyzable"] != "Yes"]
df = raw[raw["analyzable"] == "Yes"].copy()
df["reference"] = df[cfg["readers"]].mean(axis=1)
df["age_group"] = pd.cut(df["age"], [0, 39, 64, 200], labels=["18-39", "40-64", "65+"]).astype(str)
ai, ref = df[cfg["ai"]].to_numpy(), df["reference"].to_numpy()

print(f"Enrolled / Kayıtlı: {len(raw)}   Excluded / Dışlanan: {len(excl)}   Analyzed / Analiz edilen: {len(df)}")
display(excl["exclusion_reason"].value_counts().rename("n").to_frame()
        .rename_axis("Exclusion reason / Dışlanma nedeni"))
print(f"Reference range / Referans aralığı: {ref.min():.1f} – {ref.max():.1f} {U} (median/medyan {np.median(ref):.1f})")
display(df.head())
````

## 2. Primary endpoint / Birincil sonlanım noktası

**EN:** Bias = mean difference; 95 % LoA = bias ± 1.96 × SD. *Strict* success requires the 95 % CI of each LoA to lie inside the limit as well (the approach in Philips K260667).

**TR:** Bias = farkların ortalaması; %95 LoA = bias ± 1.96 × SD. *Sıkı* başarı için her LoA'nın %95 güven aralığının da sınırın içinde kalması gerekir (Philips K260667'deki yaklaşım).

````python
pr = cfg["primary"]
d1 = differences(ai, ref, pr["scale"])
r1 = bland_altman(d1)
a1 = acceptance(r1, pr["limit"])
pb1 = proportional_bias(ai, ref, d1)
sw = stats.shapiro(d1)
outside = (d1 < r1["loa_lo"]) | (d1 > r1["loa_hi"])

display(Markdown(f"**{pr['name']}** — limit / sınır ±{pr['limit']:g}"))
display(res_table(r1))
print("DECISION / KARAR:", verdict(a1))
print(f"Proportional bias / Orantısal hata: slope/eğim={pb1['slope']:+.4f}, p={pb1['p']:.4f}",
      "-> significant / anlamlı" if pb1["p"] < 0.05 else "-> not significant / anlamlı değil")
print(f"Normality of differences / Farkların normalliği (Shapiro-Wilk): W={sw.statistic:.3f}, p={sw.pvalue:.4f}")
if sw.pvalue < 0.05:
    q = np.percentile(d1, [2.5, 97.5])
    print(f"  -> Non-normal: nonparametric LoA / Normal değil: nonparametrik LoA (2.5–97.5 p): {q[0]:+.2f} / {q[1]:+.2f}")
print(f"Cases outside LoA / LoA dışı vaka: {outside.sum()} / {len(d1)} ({outside.mean()*100:.1f}%; expected/beklenen ≈ 5%)")

plot_ba(ai, ref, pr["scale"], r1, pr["limit"],
        f"{cfg['title']}\nPrimary Bland-Altman / Birincil Bland-Altman (n={r1['n']})",
        ylabel_for(pr["scale"]), "01_bland_altman_primary.png", highlight=outside)
````

**How to read / Nasıl okunur:**
- **EN:** Blue line = bias (systematic shift). Orange dashed lines = LoA, with shaded 95 % CIs. Grey band = acceptance limit. A sloped trend line means proportional error. Red circles = cases outside the LoA (≈ 5 % expected).
- **TR:** Mavi çizgi = bias (sistematik sapma). Turuncu kesikli çizgiler = LoA, gölgeli alanlar %95 GA. Gri bant = kabul sınırı. Eğimli trend çizgisi orantısal hata demektir. Kırmızı halkalar = LoA dışı vakalar (≈ %5 beklenir).

## 3. Why correlation is not enough / Neden korelasyon yetmez

**EN:** A high Pearson r shows that the two methods rise together, not that they give the same value.

**TR:** Yüksek Pearson r iki yöntemin birlikte arttığını gösterir, aynı değeri verdiğini değil.

````python
cut = cfg["cutoff"]["value"]
fig, ax = plt.subplots(figsize=(7, 7))
lo, hi = min(ai.min(), ref.min()), max(ai.max(), ref.max())
ax.plot([lo, hi], [lo, hi], color=COL["grey"], lw=1, ls="--", label="Identity / Özdeşlik (y = x)")
ax.scatter(ref, ai, s=26, color=COL["ink"], alpha=0.75)
ax.axvline(cut, color=COL["accent"], lw=1, ls=":")
ax.axhline(cut, color=COL["accent"], lw=1, ls=":", label=f"Clinical cut-off / Klinik eşik {cut:g} {U}")
r_p = stats.pearsonr(ref, ai)[0]
ax.set_title(f"{cfg['title']}\nPearson r = {r_p:.3f} (not sufficient for agreement / uyum için yeterli değil)",
             loc="left", fontsize=10)
ax.set_xlabel(f"Reference: mean of 3 experts / Referans: 3 uzman ortalaması ({U})")
ax.set_ylabel(f"AI ({U})"); ax.set_aspect("equal"); ax.grid(alpha=0.25)
ax.legend(fontsize=9, frameon=False)
fig.tight_layout(); save(fig, "02_scatter_ai_vs_reference.png"); plt.show()
````

## 4. Secondary endpoint / İkincil sonlanım noktası

**EN:** The same analysis on the other scale (descriptive; no pre-specified limit). For bladder, percent difference is restricted to reference ≥ 100 mL because percentages inflate at very small volumes.

**TR:** Aynı analiz diğer ölçekte (tanımlayıcı; önceden sınır yok). Mesanede yüzde fark yalnızca referans ≥ 100 mL vakalarda hesaplanır, çünkü çok küçük hacimlerde yüzde aşırı şişer.

````python
se = cfg["secondary"]
mask = np.ones(len(df), bool) if not se.get("min_ref") else ref >= se["min_ref"]
r2 = bland_altman(differences(ai[mask], ref[mask], se["scale"]))
a2 = acceptance(r2, se["limit"])
display(Markdown(f"**{se['name']}**"))
display(res_table(r2))
print("DECISION / KARAR:", verdict(a2))
plot_ba(ai[mask], ref[mask], se["scale"], r2, se["limit"],
        f"{cfg['title']}\nSecondary / İkincil: {se['name']} (n={r2['n']})",
        ylabel_for(se["scale"]), "03_bland_altman_secondary.png")
````

## 5. ICC, MAE and clinical cut-off / ICC, MAE ve klinik eşik

**EN:** Supportive measures. The decision is still made with the LoA versus the acceptance limit.

**TR:** Destekleyici ölçüler. Karar yine LoA ile kabul sınırına göre verilir.

````python
icc = icc_a1(ai, ref)
mae = np.mean(np.abs(ai - ref))
cat = categorical(ai, ref, cut)
display(pd.DataFrame([{"ICC(A,1)": icc, "Pearson r": r_p, f"MAE ({U})": mae}]))

display(Markdown(f"**{cfg['cutoff']['label']}**"))
cm = pd.DataFrame([[cat["tp"], cat["fn"]], [cat["fp"], cat["tn"]]],
                  index=["Reference/Referans +", "Reference/Referans −"],
                  columns=["AI +", "AI −"])
display(cm)
display(pd.DataFrame([
    {"Metric / Ölçü": "Sensitivity / Duyarlılık", "Value / Değer": cat["sens"][0],
     "95% CI / GA (Wilson)": f"{cat['sens'][1]:.3f} – {cat['sens'][2]:.3f}"},
    {"Metric / Ölçü": "Specificity / Özgüllük", "Value / Değer": cat["spec"][0],
     "95% CI / GA (Wilson)": f"{cat['spec'][1]:.3f} – {cat['spec'][2]:.3f}"},
    {"Metric / Ölçü": "Overall agreement / Toplam uyum", "Value / Değer": cat["agreement"], "95% CI / GA (Wilson)": ""},
    {"Metric / Ölçü": "Cohen kappa", "Value / Değer": cat["kappa"], "95% CI / GA (Wilson)": ""},
]).style.format({"Value / Değer": "{:.3f}"}).hide(axis="index"))
````

## 6. AI vs. human readers / AI ve insan okuyucular

**EN:** The AI's LoA is compared with how much the readers disagree with each other. If the AI's interval is no wider than the readers', the AI disagrees with experts no more than experts disagree among themselves.

**TR:** AI'ın LoA'sı, okuyucuların birbirinden ne kadar farklı ölçtüğüyle kıyaslanır. AI'ın aralığı okuyucularınkinden geniş değilse, AI uzmanlardan uzmanların birbirinden farklılaştığından daha fazla farklılaşmıyor demektir.

````python
rd = cfg["readers"]; bench = []
for i in range(len(rd)):
    for j in range(i + 1, len(rd)):
        rr = bland_altman(differences(df[rd[i]], df[rd[j]], pr["scale"]))
        bench.append({"label": f"Reader/Okuyucu {i+1} − {j+1}", **rr, "is_ai": False})
for i in range(len(rd)):
    others = df[[c for k, c in enumerate(rd) if k != i]].mean(axis=1)
    rr = bland_altman(differences(df[rd[i]], others, pr["scale"]))
    bench.append({"label": f"Reader/Okuyucu {i+1} − mean of others/diğerleri", **rr, "is_ai": False})
bench.append({"label": "AI − mean of 3 readers / 3 okuyucu ort.", **r1, "is_ai": True})
bench = pd.DataFrame(bench)
bench["LoA width / genişlik"] = bench["loa_hi"] - bench["loa_lo"]
display(bench[["label", "n", "bias", "loa_lo", "loa_hi", "LoA width / genişlik"]])
plot_intervals(bench, pr["limit"],
               "AI vs. readers: bias (dot) and 95% LoA (line) / AI ve okuyucular: bias (nokta) ve %95 LoA (çizgi)",
               ylabel_for(pr["scale"]), "04_ai_vs_human_loa.png")
````

## 7. Subgroups / Alt gruplar

**EN:** Exploratory: they do not change the study decision, but FDA expects to see consistency across site, device, sex, BMI and age. Subgroups whose LoA exceeds the limit are flagged.

**TR:** Keşifseldir: çalışma kararını değiştirmez, ama FDA merkez, cihaz, cinsiyet, BMI ve yaş genelinde tutarlılık görmek ister. LoA'sı sınırı aşan alt gruplar işaretlenir.

````python
SUBGROUP_COLS = ["site", "device_type", "sex", "bmi_category", "age_group", "image_quality"]
rows = []
for col in SUBGROUP_COLS:
    for lvl, g in df.groupby(col):
        if len(g) >= 5:
            gr = bland_altman(differences(g[cfg["ai"]], g["reference"], pr["scale"]))
            flag = (pr["limit"] is not None and (gr["loa_lo"] < -pr["limit"] or gr["loa_hi"] > pr["limit"]))
            rows.append({"label": f"{col}: {lvl}", **gr,
                         "Exceeds limit / Sınırı aşıyor": "⚠️" if flag else ""})
subs = pd.DataFrame(rows)
display(subs[["label", "n", "bias", "loa_lo", "loa_hi", "Exceeds limit / Sınırı aşıyor"]])
plot_intervals(subs, pr["limit"], "Subgroups: bias and 95% LoA (exploratory) / Alt gruplar: bias ve %95 LoA (keşifsel)",
               ylabel_for(pr["scale"]), "05_subgroup_forest.png")
````

## 8. Sample size planning / Örneklem büyüklüğü planlama

**EN:** Simulation: for each n, the probability that both LoA 95 % CIs fall inside the limit (strict criterion), given an assumed bias and SD. In a real study, take the SD from pilot data and add a 10–20 % exclusion margin.

**TR:** Simülasyon: her n için, varsayılan bias ve SD ile her iki LoA'nın %95 GA'sının sınır içinde kalma olasılığı (sıkı kriter). Gerçek çalışmada SD pilot veriden alınır ve %10–20 dışlanma payı eklenir.

````python
def simulate_power(bias, sd, delta, n, nsim=1500, rng=None):
    rng = rng or np.random.default_rng(1); ok = 0
    for _ in range(nsim):
        r = bland_altman(rng.normal(bias, sd, n))
        ok += r["loa_lo_ci_lo"] >= -delta and r["loa_hi_ci_hi"] <= delta
    return ok / nsim

if RUN_POWER:
    p = cfg["power"]; rng = np.random.default_rng(7)
    ns = np.arange(20, 401, 20)
    pw = [simulate_power(p["bias"], p["sd"], p["delta"], n, 1500, rng) for n in ns]
    fig, ax = plt.subplots(figsize=(9, 5.5))
    ax.plot(ns, pw, "o-", color=COL["ink"])
    ax.axhline(0.8, color=COL["accent"], ls="--", lw=1.2, label="80% power / güç")
    ax.axhline(0.9, color=COL["blue"], ls="--", lw=1.2, label="90% power / güç")
    ax.set_ylim(0, 1.02)
    ax.set_xlabel("Analyzable cases / Analiz edilebilir vaka (n)")
    ax.set_ylabel("P(LoA 95% CI within limit / sınır içinde)")
    ax.set_title(f"Sample size / Örneklem: bias={p['bias']}, SD={p['sd']}, limit/sınır=±{p['delta']}",
                 loc="left", fontsize=11)
    ax.grid(alpha=0.25); ax.legend(frameon=False)
    fig.tight_layout(); save(fig, "06_sample_size_power.png"); plt.show()
    for target in (0.8, 0.9):
        n_ok = next((n for n, v in zip(ns, pw) if v >= target), None)
        print(f"{int(target*100)}% power / güç -> minimum n: "
              + (str(n_ok) if n_ok else "not reached up to 400 / 400'e kadar ulaşılamadı"))
else:
    print("Skipped (RUN_POWER = False) / Atlandı (RUN_POWER = False)")
````
